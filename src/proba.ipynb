# TRANSLATION TEST

```python
import requests

def translate_apertium(url, text, langpair):
    """
    Translate text using Apertium API.
    
    params:
        url: Apertium API URL
        text: Text to be translated.
        langpair: Language pair (e.g., 'eu|en' for Basque to English).
    
    return: 
        Translated text.
    """    
    # Request parameters
    params = {
        'langpair': langpair,  # Language pair (e.g., 'eu|en')
        'q': text              # The text to be translated
    }
    
      # Make the request to the Apertium API
    response = requests.get(url, params=params)
    
    # Check if the request was successful
    if response.status_code == 200:
        # Parse the response JSON and extract the translated text
        response_json = response.json()
        translated_text = response_json.get('responseData', {}).get('translatedText', '')
        return translated_text
    else:
        # In case of an error, return the error code
        return f"Translation error: {response.status_code}"

# Apertium API URL
url = "https://www.apertium.org/apy/translate"
    
# Translation examples:
input_text_eu = "azaldu adimen artifiziala"
langpair = "eu|en"  # Basque to English
translated_text_en = translate_apertium(url, input_text_eu, langpair)
print(f"Text in Basque: {input_text_eu}")
print(f"Translated text in English: {translated_text_en}")
```

# MODEL ITERATION EXAMPLE

```python
# Test if model is working correctly with translation
import requests
from langchain.llms import Ollama

# Apertium API URL
url = "https://www.apertium.org/apy/translate"

# fuction to translate text
def translate_apertium(url, text, langpair):
    """
    Translate text using Apertium API.
    
    params:
        url: Apertium API URL
        text: Text to be translated.
        langpair: Language pair (e.g., 'eu|en' for Basque to English).
    
    return: 
        Translated text.
    """    
    # Request parameters
    params = {
        'langpair': langpair,  # Language pair (e.g., 'eu|en')
        'q': text              # The text to be translated
    }
    
      # Make the request to the Apertium API
    response = requests.get(url, params=params)
    
    # Check if the request was successful
    if response.status_code == 200:
        # Parse the response JSON and extract the translated text
        response_json = response.json()
        translated_text = response_json.get('responseData', {}).get('translatedText', '')
        return translated_text
    else:
        # In case of an error, return the error code
        return f"Translation error: {response.status_code}"

# prompt text
input_text = "Adimen artifiziala zer da?"

# Translate user text into english
langpair = "eu|en"  # Basque to English
translated_input_text = translate_apertium(url, input_text, langpair)

# Load Ollama model
llm = Ollama(model="llama3.2:1b")

# Ask to the model with tranlated text
response = llm.invoke(translated_input_text)

# check translations
print("Original prompt text:", input_text)
print("Prompt text in english:", translated_input_text)
print("Original response text:", response)
```

# CREATE WELCOME PROMPT TEMPLATE

```python
from langchain.prompts import PromptTemplate
from langchain.llms import Ollama

# Creamos un PromptTemplate con la bienvenida y las instrucciones
prompt = PromptTemplate(
    template="""
    Sukaldean lagunduko dizut. Zein ekitaldi ospatzen duzu? Zer prestatu nahi duzu? kaixo!.
    """
)

# Carga el modelo de Ollama
llm = Ollama(model="llama3.2:1b")

# Utiliza el método .format() para procesar la plantilla (si es necesario)
formatted_prompt = prompt.format()

# Translate user text into english
langpair = "eu|en"  # Basque to English
translated_input_text = translate_apertium(url, formatted_prompt, langpair)
print(translated_input_text)

# invoque welcome message
response = llm.invoke(translated_input_text)
#print(f"Original response : {response}")

# Find the position of the first period (".")
first_point = response.find(".")

# If there is a period, truncate the response up to that point, otherwise, keep the full response
if first_point != -1:
    truncated_response = response[:first_point + 1]  # Include the period
else:
    truncated_response = response  # If no period, keep the full response

# Print the truncated response
print(f"Truncated Response: {truncated_response}")
```
